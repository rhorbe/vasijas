```python
import pandas as pd
```

```python
data = pd.read_csv(
    "../output/predictions/predicciones_test_model_resnet18_unificado_bs16_lr1e-3_weighted_unfreeze.csv"
)
data["Clase Ana"] = data["Clase Ana"].apply(lambda x: int(x))
data["clase_predicha"] = data["clase_predicha"].apply(
    lambda x: int(x.replace("clase", ""))
)
```

```python
from sklearn.metrics import classification_report
```

```python
data_base = data[data["borde_origen"] == "base"]
y_true = data_base["Clase Ana"].to_numpy()
y_pred = data_base["clase_predicha"].to_numpy()
print(
    classification_report(
        y_true,
        y_pred,
    )
)
```

```python
data_base = data[data["borde_origen"] == "tope"]
y_true = data_base["Clase Ana"].to_numpy()
y_pred = data_base["clase_predicha"].to_numpy()
print(
    classification_report(
        y_true,
        y_pred,
    )
)
```
