## Algunos conceptos de las RNA

#### Función de pérdida

La función de pédida o loss function es la función que utiliza la red reuronal para calcular el error de su deducción en cada iteración.

#### Gradient descent
En el contexto de aprendizaje supervisado sirve para medir qué tan cerca de los resultados esperados es lo que produce el modelo.
Lo que hace es medir el error con un conjunto de parámetros, luego modificar esos parámetros y volver a medir el error. Si el error disminuye asume que la configuración de parámetros mejora a la anterior y se debe continuar en ese sentido. Así iterativamente mientras en error siga disminuyendo.

Es la técnica que se usa para entrenar las redes neuronales.

#### Learning rate

El tamaño de los saltos que da el algoritmo en el sentido del mínimno está determinado por el _learning rate_ o tasa de aprendizaje. El valor de la tasa de aprendizaje debe ser un valor equilibrado, ni muy alto ni muy bajo.

Esto es importante porque si los pasos son muy largos, el algoritmo podría entrar en un bucle de rebotar de un lado al otro del mínimo. Mientras que si los pasos son muy pequeños, el algoritmo podría tomar mucho más tiempo en alcanzar el mínimo y se vuelve más suceptible al problema del mínimo local.

**Local mínimo:**
En el descenso de gradiente, se inicia en un punto específico de la función de pérdida definida para los pesos de la red y se intenta ir en el sentido en el cual el error disminuye. En un caso sencillo, como el de la primer figura, cualquier paso en la dirección correcta llegará indefectiblemente al mínimo posible. Pero en los problemas de la vida real ese no suele ser el caso. Normalmente se ven situaciones como en la segunda figura, en la cual el resultado se ve "atrapado" en un mínimo local, diferente al mínimo global, que es el óptimo. En estos casos, el único camino posible es escalar a un punto más alto para poder descender al mínimo global.

**Momentum:**
Es una de las técnicas más utilizadas para ayudar a la red a evitar el problema del mínimo local. Momentum es una representación del efecto físico del momentum. Cuando la tendencia del descenso de gradiente se mantienen en el mismo sentido, agregar momentum significa incrementar el tamaño de los pasos hacia el mínimo.

#### Backpropagation

Es la técnica utilizada por la red neuronal durante el entrenamiento para ajustar sus pesos internos.
Implica calcular el error de salida y propagar hacia atrás el resultado a través de todas las conexiones de todas las capas hasta los nodos de entrada. De esa manera se distribuye el error obtenido con los pesos actuales a través de todos los nodos de la red.

Luego esa información la va a utilizar la RNA para ajustar los pesos aplicando la técnica de descenso de gradiente y poder así alcanzar mejores resultados en la próxima iteración. Este proceso se repite iterativamente hasta que el sistema converge a una solución óptima o aceptable. Es decir, hasta que el error determinado por la función converge en el mínimo.

#### Función de optimización

Son diferentes algoritmos que permiten mejoras sobre la técnica de gradient descent para optimizar el cáculo del error de la red durante el entrenamiento.

Algunos ejemplos son: momentum optimization, RMDProp, y Adam (momentum optimization + RMSProp). Esta última es una de las más utilizadas actualmente porque da buenos resultados y es fácil de usar.

En Python la librería Keras provee una clase específica para cada algoritmo de optimización que se desea aplicar.

#### Overfitting

Como cualquier otro modelo de entrenamiento supervisado, las RNA también pueden sobreajustar. Existen algunas técnicas conocidas para evitar el sobreajuste en las RNA como el _early stopping_ y el _dropout_.

**Early stopping:** significa simplemente detectar en qué momento empieza a disminuir la performance de la red y detener el entrenamiento.

**Dropout:** consiste en ignorar el resultado de un porcentaje específico de los nodos seleccionadas de manera aleatoria en cada una de las iteraciones de entrenamiento. Esto hace que el procesamiento e influencia de los nodos dentro de la red sea mejor distribuido y la red pueda generalizar mejor. Se ha probado que da muy buenos resultados.
