```python
import os
import pandas as pd

# Rutas
ruta_base = "/content/drive/MyDrive/Tesis/vasijas/Original"
archivo_csv = "vasijas4.csv"

# Leer CSV
df = pd.read_csv(os.path.join(ruta_base, archivo_csv))

# Eliminar columna archivo_encontrado si existe
if "archivo_encontrado" in df.columns:
    df = df.drop(columns=["archivo_encontrado"])

# Construir un diccionario: archivo.png -> ubicacion
archivo_a_ubicacion = {}

for ubicacion in os.listdir(ruta_base):
    ruta_ubicacion = os.path.join(ruta_base, ubicacion)
    if os.path.isdir(ruta_ubicacion):
        for archivo in os.listdir(ruta_ubicacion):
            if archivo.lower().endswith(".png"):
                archivo_a_ubicacion[archivo] = ubicacion

# Función para obtener la ubicación
def obtener_ubicacion(identificador):
    nombre_archivo = f"{identificador}.png"
    return archivo_a_ubicacion.get(nombre_archivo)

# Agregar columna ubicacion
df["Ubicacion"] = df["Identificador"].astype(str).apply(obtener_ubicacion)

# Guardar nuevo archivo
archivo_salida = os.path.join(ruta_base, "vasijas5.csv")
df.to_csv(archivo_salida, index=False)

print("Archivo generado:", archivo_salida)
```
