```python
from pathlib import Path
import shutil
import pandas as pd
from _bootstrap import setup

def copiar_perfiles_a_unificado(ruta_origen: Path, ruta_unificado: Path, origen_label: str) -> list[dict]:
    filas = []
    for p in ruta_origen.rglob("*.png"):
        # ruta relativa dentro de la estructura de clases (ej. clase01/archivo.png)
        try:
            rel = p.relative_to(ruta_origen)
        except Exception:
            # fallback: usar nombre solo
            rel = Path(p.name)
        destino = ruta_unificado / rel
        destino.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(p, destino)
        filas.append({
            "archivo_generado": destino.name,
            "archivo_origen": p.name,
            "origen": origen_label,
            "clase_carpeta": destino.parent.name
        })
    return filas

# ejecución
PATHS = setup()
ruta_unificado = PATHS.DATA_UNIFICADO
ruta_unificado.mkdir(parents=True, exist_ok=True)

filas_totales = []
filas_totales += copiar_perfiles_a_unificado(PATHS.DATA_CENTRADOS, ruta_unificado, "centrados")
filas_totales += copiar_perfiles_a_unificado(PATHS.DATA_MITADES, ruta_unificado, "mitades")

df_unificado = pd.DataFrame(filas_totales)
csv_destino = ruta_unificado / "metadata.csv"
df_unificado.to_csv(csv_destino, index=False)

print(f"Copiados {len(df_unificado)} archivos a {ruta_unificado}")
```
