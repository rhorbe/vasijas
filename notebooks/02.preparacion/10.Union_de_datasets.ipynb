# Unión de datasets

La lógica de copia, unificación y generación de metadata vive en `vasijas.datasets`. Este notebook solo define las fuentes y ejecuta el proceso.


```python
from vasijas.datasets import DatasetFuente, unir_datasets_png
from vasijas.paths import setup
```

```python
PATHS = setup()

DATASETS_FUENTE = [
    DatasetFuente(PATHS.DATA_CENTRADOS, "centrados"),
    DatasetFuente(PATHS.DATA_MITADES, "mitades"),
]
```

```python
resultado = unir_datasets_png(
    datasets_fuente=DATASETS_FUENTE,
    ruta_destino=PATHS.DATA_UNIFICADO,
)

resultado.mensaje_resumen
```
