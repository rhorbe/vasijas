# Unión de datasets

## Unión de datasets de imágenes cuadradas y centradas


```python
import pandas as pd
from vasijas.config import load_config
from vasijas.metadata import (
    cargar_metadata,
    completar_metadata_cuadrados,
    get_columnas_adicionales_metadata,
    guardar_metadata,
)
from vasijas.paths import ProjectPaths
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)
```

## Unificación de formato de metadata


### Metadata de perfiles completos

```python
df_metadata_cuadrados = cargar_metadata(PATHS.DATA_CUADRADOS)
display(df_metadata_cuadrados.head())
```

### Metadata de fracciones de perfiles

```python
df_metadata_centrados = cargar_metadata(PATHS.DATA_CENTRADOS)
display(df_metadata_centrados.head())
```

### Columnas que están en el metadata de las fracciones que no están en el metadata de los perfiles completos

```python
columnas_adicionales_metadata = get_columnas_adicionales_metadata(
    df_metadata_cuadrados,
    df_metadata_centrados,
)
for columna in columnas_adicionales_metadata:
    print(columna)
```

```python
df_metadata_cuadrados_completo = completar_metadata_cuadrados(
    df_metadata_cuadrados=df_metadata_cuadrados,
    df_metadata_referencia=df_metadata_centrados,
    columnas_derivadas=columnas_adicionales_metadata,
)

df_metadata_cuadrados_completo.head()
```

```python
assert list(df_metadata_cuadrados_completo.columns) == list(
    df_metadata_centrados.columns
)

df_metadata_unificados = pd.concat(
    [df_metadata_cuadrados_completo, df_metadata_centrados],
    ignore_index=True,
)

assert len(df_metadata_unificados) == len(df_metadata_cuadrados_completo) + len(
    df_metadata_centrados
)

df_metadata_unificados.head()
```

```python
path_metadata_unificado = guardar_metadata(df_metadata_unificados, PATHS.DATA_UNIFICADO)
print(f"El nuevo archivo metadata ha sido guardado en: {path_metadata_unificado}")
```

## Unión de imágenes PNG


```python
from vasijas.ubicacion import copiar_estructura_archivos

cant_archivos_centrados_copiados = copiar_estructura_archivos(
    PATHS.DATA_CENTRADOS, PATHS.DATA_UNIFICADO
)
cant_archivos_cuadrados_copiados = copiar_estructura_archivos(
    PATHS.DATA_CUADRADOS, PATHS.DATA_UNIFICADO
)

total_archivos_copiados = (
    cant_archivos_centrados_copiados + cant_archivos_cuadrados_copiados
)

print(f"Se copiaron un total de: {total_archivos_copiados}")
```
