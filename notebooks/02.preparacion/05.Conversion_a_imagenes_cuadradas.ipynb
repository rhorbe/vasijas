# Conversión de perfiles a imágenes cuadradas

Las imágenes de los perfiles tienen distintas formas rectangulares, en esta sección se las convierte en imágenes cuadradas.

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.imagenes import convertir_dataset_clasificado_a_cuadrado
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CUADRAR = [*range(1, 12), "sin_clase"]
```

```python
resultado = convertir_dataset_clasificado_a_cuadrado(
    ruta_origen=PATHS.DATA_CLASIFICADOS,
    ruta_destino=PATHS.DATA_CUADRADOS,
    clases=CLASES_A_CUADRAR,
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
```
