# Corte en mitades

En esta sección, las imágenes de los perfiles se dividen en dos mitades: una superior, denominada tope, y una inferior, denominada base.

Como resultado de este proceso, por cada perfil original se generan dos nuevas imágenes. El nombre de cada archivo conserva el identificador del perfil original y añade el sufijo _tope o _base, según corresponda a la mitad superior o inferior.

Paralelamente, se genera un nuevo archivo de metadatos. Al igual que en el conjunto de datos original, el campo identificador coincide con el nombre del archivo de imagen correspondiente.

Además, se incorporan las siguientes columnas:

* *id_perfil_completo*: conserva la referencia al perfil original del cual se obtuvo cada mitad.
* *borde_original*: indica el extremo de origen del fragmento, pudiendo tomar los valores tope para la mitad superior o base para la mitad inferior.
* *porcentaje_avance*: indica la proporción del perfil conservada a partir del borde de origen. Dado que en esta etapa cada perfil se divide exactamente en dos mitades, este valor es siempre del 50 %.

```python
from vasijas.config import load_config
from vasijas.metadata import cargar_metadata
from vasijas.paths import ProjectPaths
from vasijas.pipelines import cortar_dataset_clasificado_en_mitades
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CORTAR = [4]

DATA_ORIGEN = PATHS.DATA_CLASIFICADOS
DATA_DESTINO = PATHS.DATA_MITADES
```

```python
resultado = cortar_dataset_clasificado_en_mitades(
    ruta_origen=PATHS.DATA_CLASIFICADOS,
    ruta_destino=PATHS.DATA_MITADES,
    clases=CLASES_A_CORTAR,
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
```
