# Corte en mitades

En esta sección las imágenes de los perfiles son cortadas en mitades.

Se crea un nuevo archivo de metadata en el cual a los identificadores se les agrega el sufijo "_tope" si la imagen corresponde a la mitad superior y "_base" si corresponde a la mitad inferior.

Se agrega además una columna denominada "mitad" la cual también indica si se trata de un tope o una base.

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.imagenes import cortar_dataset_clasificado_en_mitades
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CORTAR = [*range(1, 12)]
```

```python
resultado = cortar_dataset_clasificado_en_mitades(
    ruta_origen=PATHS.DATA_CLASIFICADOS,
    ruta_destino=PATHS.DATA_MITADES,
    clases=CLASES_A_CORTAR,
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
```
