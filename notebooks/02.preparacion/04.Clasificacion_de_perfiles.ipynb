# Clasificación de perfiles

Se clasifican los perfiles según la Clase Ana, octava columna del archivo de metadata.

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths, ruta_relativa
from vasijas.archivos import copiar_perfiles_por_clase
from vasijas.metadata import cargar_metadata, copiar_metadata
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)
```

```python
# Configuración
COL_ID = "Identificador"
COL_CLASE = "Clase Ana"
COL_UBICACION = "Ubicacion"
```

```python
# Ejecución
df = cargar_metadata(PATHS.DATA_UBICADOS)

# Visualización
display(df.head())
```

```python
resultado_clasificacion = copiar_perfiles_por_clase(
    df=df,
    ruta_origen=PATHS.DATA_UBICADOS,
    ruta_destino=PATHS.DATA_CLASIFICADOS,
    columna_identificador=COL_ID,
    columna_clase=COL_CLASE,
    columna_ubicacion=COL_UBICACION,
)

for ruta_no_encontrada in resultado_clasificacion.rutas_no_encontradas:
    print(f"No se encontró el archivo: {ruta_relativa(ruta_no_encontrada)}")

print(f"\nArchivos copiados: {resultado_clasificacion.archivos_copiados}")
print(f"Archivos no encontrados: {resultado_clasificacion.archivos_no_encontrados}\n")

print("Resumen de archivos copiados por clase:")
for carpeta, cantidad in sorted(resultado_clasificacion.conteo_por_clase.items()):
    print(f"{ruta_relativa(carpeta)}: {cantidad}")
```

```python
metadata_copiada = copiar_metadata(
    ruta_origen=PATHS.DATA_UBICADOS,
    ruta_destino=PATHS.DATA_CLASIFICADOS,
)
display(ruta_relativa(metadata_copiada))
```
