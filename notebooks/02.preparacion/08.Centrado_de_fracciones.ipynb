# Centrado de fracciones


Las fracciones de los perfiles generados en la sección anterior son centrados en la imagen cuadrada.

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.imagenes import centrar_imagenes
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CENTRAR = [1]
NOMBRE_METADATA = "metadata.csv"
```

```python
resultado = centrar_imagenes(
    ruta_origen=PATHS.DATA_FRACCIONES,
    ruta_destino=PATHS.DATA_CENTRADOS,
    clases=CLASES_A_CENTRAR,
)

resultado.mensaje_resumen
```
