# Centrado de fracciones


En esta sección, las imágenes correspondientes a las fracciones de los perfiles generadas en la etapa anterior se centran dentro de una imagen cuadrada.

Como resultado de este proceso, se genera un nuevo archivo de imagen para cada fracción centrada y se actualiza el archivo de metadatos para reflejar los cambios realizados.

Al nombre de cada archivo se le agrega el sufijo _centrado, con el fin de diferenciar la imagen procesada de la fracción original. Por ejemplo, el archivo _al_13b_tope_10_centrado.png_ corresponde al perfil al_13b, conservando un 10 % de su extensión a partir del borde superior (tope) y con la fracción resultante centrada dentro de la imagen cuadrada.

El identificador de cada fracción centrada coincide con el nombre del archivo de imagen correspondiente, siguiendo el mismo criterio utilizado en las etapas anteriores del procesamiento.

No se agregan nuevas columnas al archivo de metadatos.

```python
from vasijas.config import load_config
from vasijas.metadata import cargar_metadata
from vasijas.paths import ProjectPaths
from vasijas.pipelines import centrar_dataset_clasificado_de_fracciones
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CENTRAR = [4]

DATA_ORIGEN = PATHS.DATA_FRACCIONES
DATA_DESTINO = PATHS.DATA_CENTRADOS
```

```python
resultado = centrar_dataset_clasificado_de_fracciones(
    ruta_origen=DATA_ORIGEN,
    ruta_destino=DATA_DESTINO,
    clases=CLASES_A_CENTRAR,
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    print("Errores:")
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
else:
    print("No se produjeron errores.")
```
