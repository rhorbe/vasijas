# Corte en fracciones


En esta sección, las imágenes de los perfiles son fraccionadas a partir de uno de sus extremos. El fraccionamiento se realiza según un porcentaje de avance, que determina qué proporción del perfil se conserva desde el borde seleccionado.

Como resultado, para cada perfil se generan tantos archivos como sean posibles de acuerdo con el porcentaje de avance definido.

A cada archivo generado se le agrega un sufijo que indica el porcentaje de avance correspondiente. Por ejemplo, el archivo *al_13b_tope_10.png* corresponde al perfil *al_13b*, considerando un 10% de avance desde el borde superior, denominado *tope*.

El archivo de metadata también incluirá este nuevo identificador, respetando el mismo formato utilizado en el nombre del archivo.


```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.imagenes import cortar_dataset_clasificado_en_fracciones
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CORTAR = [*range(1, 12)]
PORCENTAJE_AVANCE = 10.0
```

```python
resultado = cortar_dataset_clasificado_en_fracciones(
    ruta_origen=PATHS.DATA_CLASIFICADOS,
    ruta_destino=PATHS.DATA_FRACCIONES,
    clases=CLASES_A_CORTAR,
    porcentaje_avance=PORCENTAJE_AVANCE,
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
```
