# Corte en fracciones


En esta sección, las imágenes de los perfiles se fraccionan a partir de uno de sus extremos: el borde superior (tope) o el borde inferior (base). El fraccionamiento se realiza según un porcentaje de avance, que determina la proporción del perfil que se conserva desde el borde seleccionado.

Como resultado, para cada perfil se generan tantos fragmentos como sea posible de acuerdo con el porcentaje de avance definido.

A cada archivo generado se le asigna un nombre que conserva el identificador del perfil original e incorpora un sufijo que indica el borde de origen y el porcentaje de avance correspondiente. Por ejemplo, el archivo al_13b_tope_10.png representa el perfil al_13b conservando un 10 % de su extensión a partir del borde superior (tope).

Paralelamente, se genera un nuevo archivo de metadatos. En este archivo, el identificador de cada fragmento coincide con el nombre del archivo de imagen correspondiente, siguiendo el mismo criterio utilizado en el conjunto de datos original.

Además, se incorporan las siguientes columnas:

* *id_perfil_completo*: conserva la referencia al perfil original del cual se obtuvo el fragmento.
* *borde_original*: indica el extremo desde el cual se generó el fragmento, pudiendo tomar los valores tope o base.
* *porcentaje_avance*: indica el porcentaje del perfil conservado desde el borde de origen.

Por lo tanto, el identificador registrado en los metadatos mantiene el mismo formato que el nombre del archivo generado y permite determinar directamente tanto el perfil de origen como el borde considerado y el porcentaje de avance aplicado.

```python
from vasijas.config import load_config
from vasijas.metadata import cargar_metadata
from vasijas.paths import ProjectPaths
from vasijas.pipelines import cortar_dataset_clasificado_en_fracciones
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)

CLASES_A_CORTAR = [4]
PORCENTAJE_AVANCE = 10.0

DATA_ORIGEN = PATHS.DATA_CLASIFICADOS
DATA_DESTINO = PATHS.DATA_FRACCIONES
```

```python
resultado = cortar_dataset_clasificado_en_fracciones(
    ruta_origen=DATA_ORIGEN,
    ruta_destino=DATA_DESTINO,
    clases=CLASES_A_CORTAR,
    porcentaje_avance=PORCENTAJE_AVANCE
)

resultado.mensaje_resumen
```

```python
if resultado.errores:
    print("Errores:")
    for error in resultado.errores:
        print(f"{error.ruta_origen}: {error.mensaje}")
else:
    print("No se produjeron errores.")
```
