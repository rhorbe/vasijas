# Agregado de ubicación a metadata

En esta sección se realiza agrega al metadata una columna con el nombre de la ubicación donde fue encontrada la pieza.

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.metadata import (
    cargar_metadata,
    guardar_metadata,
)
from vasijas.ubicacion import (
    agregar_ubicacion,
    copiar_estructura_archivos,
    mapear_archivos_a_ubicacion,
)
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)
```

## Previsualización de metadata antes de editar

```python
# Leer CSV
df = cargar_metadata(PATHS.DATA_CORREGIDOS)
display(df.head())
```

## Relación identificador - ubicación

```python
# Archivos
archivo_ubicacion: dict[str, str] = mapear_archivos_a_ubicacion(PATHS.DATA_CORREGIDOS)
# Preview
display(list(archivo_ubicacion.items())[:10])
```

## Agregado de ubicación al archivo de metadata

```python
# Agregar columna ubicacion
df, mensaje = agregar_ubicacion(df, archivo_ubicacion)
display(df.head())
print(mensaje)
```

## Copia de archivos a carpeta con el metadata actualizado

```python
# Copiar toda la estructura de DATA_CORREGIDOS a DATA_UBICADOS
archivos_copiados = copiar_estructura_archivos(
    PATHS.DATA_CORREGIDOS,
    PATHS.DATA_UBICADOS,
)

print(f"{archivos_copiados} archivos copiados.")
print(f"Desde {PATHS.DATA_CORREGIDOS}\n" f"Hacia {PATHS.DATA_UBICADOS}")
```

## Guardado de nuevo archivo de metadata

```python
# Guardar nuevo archivo metadata
ruta_metadata_ubicado = guardar_metadata(df, PATHS.DATA_UBICADOS)
print(f"Ruta metadata con ubicación {ruta_metadata_ubicado}")
```
