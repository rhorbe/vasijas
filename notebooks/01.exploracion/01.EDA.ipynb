# Análisis Exploratorio de los Datos

```python
from vasijas.config import load_config
from vasijas.paths import ProjectPaths
from vasijas.metadata import (
    cargar_metadata,
    contar_registros_metadata,
    get_identificadores_metadata,
    get_preview_metadata,
)
from vasijas.eda import (
    agregar_marca_si_id_esta_en_fs,
    contar_archivos_sin_registro,
    contar_identificadores_encontrados,
    contar_identificadores_unicos_no_encontrados,
    contar_nombres_invalidos,
    contar_perfiles_por_ubicacion,
    contar_ubicaciones,
    get_archivos_sin_registro,
    get_identificadores_no_encontrados,
    get_nombres_archivos_perfiles,
    get_nombres_invalidos,
    get_ubicaciones,
    mostrar_duplas,
    mostrar_en_columnas,
)
```

```python
CONFIG = load_config()
PATHS = ProjectPaths.from_root(CONFIG.project_root)
```

## Ubicaciones

### En los datos originales

```python
# Ubicaciones detectadas en los datos originales
ubicaciones_ordenadas = get_ubicaciones(PATHS.DATA_ORIGINALES)
cantidad_ubicaciones = contar_ubicaciones(ubicaciones_ordenadas)

print(f"Cantidad de ubicaciones: {cantidad_ubicaciones}\n")
print("Detalle de ubicaciones encontradas:")
for i, ubicacion in enumerate(ubicaciones_ordenadas, start=1):
    print(f"{i}.", ubicacion)
```

## Perfiles

### Perfiles por ubicación

```python
# Total de archivos y cantidad de perfiles encontrados por ubicación
total_png, conteo_png_ordenado = contar_perfiles_por_ubicacion(PATHS.DATA_ORIGINALES)

print(f"Total de perfiles (archivos PNG) en filesystem: {total_png}\n")
print("Detalle de perfiles por ubicación:")
mostrar_duplas(conteo_png_ordenado)
```

```python
nombres_archivos_perfiles = get_nombres_archivos_perfiles(PATHS.DATA_ORIGINALES)

print("Muestra de nombres de archivos:")
for nombre in list(nombres_archivos_perfiles)[:10]:
    print(f"  {nombre}")
```

### Verificación de formato de nombres de archivos

Se adoptó un criterio de validación para el nombre base de los archivos (sin su extensión).

Según este criterio, debía cumplir las siguientes condiciones:

* Comenzar con exactamente dos letras sin acentos, comprendidas entre A–Z o a–z.
* Continuar con un guión bajo (_).
* Incluir, a continuación, al menos dos dígitos consecutivos.
* Contener, después de esos dígitos y de manera opcional, únicamente letras sin acentos, dígitos o guiones bajos.

No se admitieron espacios, guiones medios (-), paréntesis ni otros caracteres especiales.


```python
# Verificar cuáles nombres no cumplen con el formato
nombres_invalidos_ordenados = get_nombres_invalidos(nombres_archivos_perfiles)
cantidad_invalidos = contar_nombres_invalidos(nombres_invalidos_ordenados)

print(f"Cantidad de nombres que NO cumplen el formato: {cantidad_invalidos}\n")
print("Detalle de nombres inválidos:")
mostrar_en_columnas(nombres_invalidos_ordenados, cantidad_columnas=5)
```

## Metadata

### Comparación entre metadata y filesystem

```python
# Comparación Identificadores del dataset / nombres de los archivos
df = agregar_marca_si_id_esta_en_fs(df, nombres_archivos_perfiles)
cantidad_encontrados = contar_identificadores_encontrados(df)
cantidad_no_encontrados = cantidad_registros - cantidad_encontrados

print(f"Total de registros en metadata.csv: {cantidad_registros}.")
print(f"Coincidentes con nombres de archivos: {cantidad_encontrados}.")
print(f"No encontrados: {cantidad_no_encontrados}.\n")
```

#### Identificadores del dataset que no fueron encontrados en el conjunto de nombres de archivos

```python
# Detalle de Identificadores del dataset que no encontrados en el conjunto de nombres de archivos:

ids_no_encontrados_ordenados = get_identificadores_no_encontrados(df)
cantidad_no_encontrados = contar_identificadores_unicos_no_encontrados(df)

print(
    f"Cant. de Identificadores del metadata que aparecen como nombre de un archivo PNG:{cantidad_no_encontrados}\n"
)
print("Detalle de identificadores no encontrados en archivos PNG:\n")

mostrar_en_columnas(ids_no_encontrados_ordenados, cantidad_columnas=5)
```
