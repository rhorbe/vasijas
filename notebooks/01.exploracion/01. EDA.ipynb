# Análisis Exploratorio de los Datos


```python
from src.vasijas._bootstrap import setup
from src.vasijas.eda import get_ubicaciones
from src.vasijas.files import validar_formato_nombre
```

```python
PATHS = setup()
```

```python
# Ubicaciones detectadas en los datos 01.originales

ubicaciones_ordenadas = get_ubicaciones(PATHS.DATA_ORIGINALES)
cantidad_ubicaciones = len(ubicaciones_ordenadas)

print(f"Cantidad de ubicaciones: {cantidad_ubicaciones}\n")
for i, ubicacion in enumerate(ubicaciones_ordenadas , start=1):
    print(f"{i}.", ubicacion)
```

```python
# Perfiles encontrados por ubicación

total_png = 0
conteo_png: dict[str, int] = dict()

for ruta_ubicacion in PATHS.DATA_ORIGINALES.iterdir():
    if ruta_ubicacion.is_dir():
        pngs = list(ruta_ubicacion.glob("*.png"))
        conteo_png[ruta_ubicacion.name] = len(pngs)
        total_png += len(pngs)

conteo_png_ordenado = sorted(conteo_png.items(), key=lambda x: x[1])

print(f"Total de archivos PNG (perfiles) en filesystem: {total_png}\n")
for ubicacion, conteo in conteo_png_ordenado:
    print(f"{ubicacion}: {conteo}")
```

```python
# Archivos
nombres_archivos_png: set[str] = {p.stem for p in PATHS.DATA_ORIGINALES.rglob("*.png")}

# Preview
print(f"Muestra:{list(nombres_archivos_png)[:10]}")
```

```python
# Verificar cuáles nombres no cumplen con el formato
nombres_invalidos = {nombre for nombre in nombres_archivos_png if not validar_formato_nombre(nombre)}
cantidad_invalidos = len(nombres_invalidos)

print(f"Cantidad de nombres que NO cumplen el formato: {cantidad_invalidos}\n")
print("Detalle de nombres inválidos:")
for nombre in sorted(nombres_invalidos):
    print(f"  {nombre}")
```

```python
# Dataframe
import pandas as pd

archivo_csv = PATHS.DATA_ORIGINALES / "metadata.csv"
df = pd.read_csv(archivo_csv)

print(f"Cantidad de registros en metadata.csv: {len(df)}\n")
print(f"Cantidad de registros en metadata.csv: {len(df)}\n")
print(df.head())

head_string = df.head().to_string()
```

```python
print(df.info())
```

```python
# Verificar cuáles identificadores no cumplen con el formato
nombres_invalidos = {nombre for nombre in nombres_archivos_png if not validar_formato_nombre(nombre)}
cantidad_invalidos = len(nombres_invalidos)

print(f"Cantidad de nombres que NO cumplen el formato: {cantidad_invalidos}\n")
print("Detalle de nombres inválidos:")
for nombre in sorted(nombres_invalidos):
    print(f"  {nombre}")
```

```python
# Identificadores del dataset que aparecen en los nombres de los archivos
df["id_encontrado_en_fs"] = (
    df["Identificador"]
    .astype(str)
    .str.strip()
    .isin(nombres_archivos_png)
)

cantidad_encontrados = int(df["id_encontrado_en_fs"].sum())
cantidad_no_encontrados = int((~df["id_encontrado_en_fs"]).sum())

print(f"Cantidad de Identificadores que aparecen en los nombres de los archivos: {cantidad_encontrados}")
print(f"Cantidad de Identificadores que NO aparecen en los nombres de los archivos: {cantidad_no_encontrados}\n")
```

```python
# Detalle de Identificadores del dataset que no encontrados en el conjunto de nombres de archivos:

ids_no_encontrados = set(df.loc[~df["id_encontrado_en_fs"], "Identificador"].astype(str).str.strip())
cantidad_no_encontrados = len(ids_no_encontrados)
ids_no_encontrados_ordenados = sorted(ids_no_encontrados)


print("Cant. de Identificadores que aparecen en el metadata que no se corresponde con el nombre de un archivo PNG:", cantidad_no_encontrados)
print("Detalle de identificadores no encontrados en archivos PNG:")
print(ids_no_encontrados_ordenados)
```

```python
# Archivos PNG sin registro en dataset

ids_en_df = set(df["Identificador"].astype(str).str.strip())

archivos_sin_registro = nombres_archivos_png - ids_en_df
cantidad_archivos_sin_registro = len(archivos_sin_registro)

archivos_sin_registro_ordenados = sorted(archivos_sin_registro)


print(f"Cantidad de archivos PNG sin registro en metadata: {cantidad_archivos_sin_registro}")
print("Detalle de archivos PNG no encontrados en metadata.csv:")
print(archivos_sin_registro_ordenados)
```
