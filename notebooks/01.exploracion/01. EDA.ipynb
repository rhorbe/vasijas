# Análisis Exploratorio de los Datos


```python
from src.vasijas.paths import setup
import src.vasijas.eda as eda
```

```python
PATHS = setup()
```

```python
# Ubicaciones detectadas en los datos 01.originales
ubicaciones_ordenadas = eda.get_ubicaciones(PATHS.DATA_ORIGINALES)
cantidad_ubicaciones = len(ubicaciones_ordenadas)

print(f"Cantidad de ubicaciones: {cantidad_ubicaciones}\n")
for i, ubicacion in enumerate(ubicaciones_ordenadas , start=1):
    print(f"{i}.", ubicacion)
```

```python
# Total de archivos y cantidad de perfiles encontrados por ubicación

total_png, conteo_png_ordenado = eda.contar_perfiles_por_ubicacion(PATHS.DATA_ORIGINALES)

print(f"Total de archivos PNG (perfiles) en filesystem: {total_png}\n")
for ubicacion, conteo in conteo_png_ordenado:
    print(f"{ubicacion}: {conteo}")
```

```python
nombres_archivos_perfiles = eda.get_nombres_archivos_perfiles(PATHS.DATA_ORIGINALES)

print(f"Muestra nombre de archivos:{list(nombres_archivos_perfiles)[:10]}")
```

```python
# Verificar cuáles nombres no cumplen con el formato
cantidad_invalidos, nombres_invalidos_ordenados = eda.resumir_nombres_invalidos(nombres_archivos_perfiles)

print(f"Cantidad de nombres que NO cumplen el formato: {cantidad_invalidos}\n")
print("Detalle de nombres inválidos:")
for nombre in nombres_invalidos_ordenados:
    print(f"  {nombre}")
```

```python
# Dataframe
df = eda.cargar_metadata(PATHS.DATA_ORIGINALES)
cantidad_registros, preview_metadata = eda.resumir_metadata(df)

print(f"Cantidad de registros en metadata.csv: {cantidad_registros}\n")
print(preview_metadata)
```

```python
print(df.info())
```

```python
# Identificadores del dataset que aparecen en los nombres de los archivos
df = eda.agregar_marca_identificador_en_fs(df, nombres_archivos_perfiles)
cantidad_encontrados = eda.contar_identificadores_encontrados(df)
cantidad_no_encontrados = eda.contar_identificadores_no_encontrados(df)

print(f"Cantidad de Identificadores que aparecen en los nombres de los archivos: {cantidad_encontrados}")
print(f"Cantidad de Identificadores que NO aparecen en los nombres de los archivos: {cantidad_no_encontrados}\n")
```

```python
# Detalle de Identificadores del dataset que no encontrados en el conjunto de nombres de archivos:

ids_no_encontrados_ordenados = eda.get_identificadores_no_encontrados(df)
cantidad_no_encontrados = eda.contar_identificadores_unicos_no_encontrados(df)

print("Cant. de Identificadores que aparecen en el metadata que no se corresponde con el nombre de un archivo PNG:", cantidad_no_encontrados)
print("Detalle de identificadores no encontrados en archivos PNG:")
print(ids_no_encontrados_ordenados)
```

```python
# Archivos PNG sin registro en dataset

ids_metadata = eda.get_identificadores_metadata(df)
archivos_sin_registro_ordenados = eda.get_archivos_sin_registro(nombres_archivos_perfiles, ids_metadata)
cantidad_archivos_sin_registro = eda.contar_archivos_sin_registro(nombres_archivos_perfiles, ids_metadata)

print(f"Cantidad de archivos PNG sin registro en metadata: {cantidad_archivos_sin_registro}")
print("Detalle de archivos PNG no encontrados en metadata.csv:")
print(archivos_sin_registro_ordenados)
```
