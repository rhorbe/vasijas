```python
import os
import glob

conteo_png = {}
total_png = 0

for ubicacion in os.listdir(ruta_base):
    ruta_ubicacion = os.path.join(ruta_base, ubicacion)
    if os.path.isdir(ruta_ubicacion):
        pngs = glob.glob(os.path.join(ruta_ubicacion, "*.png"))
        conteo_png[ubicacion] = len(pngs)
        total_png += len(pngs)

print (f"Total de perfiles: {total_png}")
conteo_png
```

Creación de un nuevo archivo

```python
import os
import pandas as pd

# -----------------------------
# Configuración
# -----------------------------
ruta_base = "/content/drive/MyDrive/Tesis/vasijas/Original/"
archivo_csv = os.path.join(ruta_base, "vasijas.csv")
archivo_salida = os.path.join(ruta_base, "vasijas2.csv")
```

```python
# -----------------------------
# 1. Obtener TODOS los archivos existentes en las carpetas de ubicaciones
# -----------------------------
archivos_existentes = set()

for carpeta in os.listdir(ruta_base):
    ruta_carpeta = os.path.join(ruta_base, carpeta)
    if os.path.isdir(ruta_carpeta):
        for archivo in os.listdir(ruta_carpeta):
            archivos_existentes.add(archivo)

archivos_existentes
```

```python
# -----------------------------
# 2. Leer el CSV original
# -----------------------------
df = pd.read_csv(archivo_csv)

```

```python
# -----------------------------
# 3. Verificar existencia del archivo
# -----------------------------
df["archivo_encontrado"] = (
    df["Identificador"]
    .astype(str)
    .str.strip()
    .add(".png")
    .isin(archivos_existentes)
)
df.head()
```

```python
# -----------------------------
# 4. Guardar nuevo CSV
# -----------------------------
df.to_csv(archivo_salida, index=False)

print(f"Archivo generado correctamente: {archivo_salida}")
```

```python
# -----------------------------
# 5. Listar archivos no encontrados
# -----------------------------
df_no_encontrados = df[df["archivo_encontrado"] == False]
df_no_encontrados
```

Cotejando los archivos que en vasijas2.csv aparecen como no encontrados y la ubicación en las carpetas, se puede ver que los archivos existen pero hay errores en la escritura del campo Identificador.

Se procede a unificar criterios:
* Todos los identificadores y nombres de archivos estarán escritos en letras mayúsculas.
  * Se cambian los identificadores Al_03G	y Al_06A por AL_03G	y AL_06A.
* Las primeras letras de los identicadores y los archivos referencian al lugar. Por ejemplo AL hace referencia a Almedinilla. Estas letras deben estar seguidas por un guión bajo antes del número de pieza.
  * Se cambian los identificadores:
    * BA115_5 por BA_115_5
    * BA132_1 por BA_132_1.
    * PN10_1 por PN_10_1.
    * CC0_8_2 por CC_0_8_2.
  * Se renombran los archivos de La Bobadilla, los cuales comienzan con el prefijo BO, agregando un guión bajo luego del mismo. Por ejemplo, BO3_1.png se cambia por BO_3_1.png.
  * En el caso puntual del identificador BO3_2_1 se deduce que hay un error de tipeo y que en realidad se trata del archivo BO_32_1.png dado que este último no está identificado. Además al observar los archivos de la carpeta "La Bobadilla" el nombre BO_32_1.png guarda correlación numérica y de formato con los mismos por lo que de asume que el error está en el identificador.
* Los nombres de los archivos no deber contener espacios en blanco.
  * Se reemplazan por guiones bajos los espacios en los nombres de los archivos:
    * CC_13_14_1_3
    * CC_13_14_2_1
    * CC_13_14_2_3

Los cambios en los identificadores se realizan manualmente editando el archivo en Hojas de Cálculo de Google.

El renombramiento de los archivos que contienen espacios también se realiza manualmente. Los de la carpeta La Bobadilla, a los cuales se les debe agregar un guión bajo se hace mediante el código que se presenta a continuación:

```python
# Renombrado de los archivos de La Bobadilla
ruta_archivos_bobadilla = ruta_base + "/La Bobadilla"

for nombre in os.listdir(ruta_archivos_bobadilla):
    if nombre.startswith("BO") and not nombre.startswith("BO_"):
        origen = os.path.join(ruta_archivos_bobadilla, nombre)
        destino = os.path.join(ruta_archivos_bobadilla, "BO_" + nombre[2:])
        os.rename(origen, destino)
```
